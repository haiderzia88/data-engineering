## 1. What Is a Variable?

A variable name refers to a value. Use `=` to assign a value to a name.
Python evaluates the expression on the right before assigning its result
to the name on the left.

Here, the variable names explain what each number represents:

````python
apples = 6
oranges = 4
total_fruit = apples + oranges

print("Apples:", apples)
print("Oranges:", oranges)
print("Total fruit:", total_fruit)
````

`apples` refers to the variable's value. `"apples"` is literal text.
Quotes change the meaning:

````python
apples = 6

print(apples)     # Show the value stored under this name.
print("apples")   # Show the text inside the quotes.
````

## 2. Choose Clear Variable Names

Use descriptive `snake_case` names: lowercase words separated by underscores.
For these examples, use letters, digits, and underscores, starting with
a letter or underscore.

| Good name | What it describes |
| --- | --- |
| `student_name` | A student's name |
| `books_read` | The number of books read |
| `is_finished` | Whether an activity is finished |

- Names are case sensitive: `age` and `Age` are different names.
- Names cannot contain spaces or hyphens, or start with a digit.
- Reserved words such as `class` cannot be variable names.
- Avoid replacing built-in names such as `print`, `int`, or `str`.

## 3. Understand the Basic Types

A value's type determines the operations that make sense for it.

| Type | Example | Typical use |
| --- | --- | --- |
| `int` | `24` | Whole numbers, such as ages or counts |
| `float` | `1.75` | Measurements or decimal quantities |
| `str` | `"Amina"` | Text |
| `bool` | `True` | True or false flags |
| `NoneType` | `None` | A value representing absence |

`True`, `False`, and `None` must use the capitalization shown.
`"False"` and `"None"` are strings because they are quoted.

Use `type(value)` to inspect a value's type:

````python
student_name = "Amina"
age = 24
height_metres = 1.65
is_student = True
middle_name = None

print("student_name:", type(student_name))
print("age:", type(age))
print("height_metres:", type(height_metres))
print("is_student:", type(is_student))
print("middle_name:", type(middle_name))
````

`None` means that no value has been supplied in this example. It is
different from `0`, `False`, and an empty string (`""`).

## 4. Update Variables

Assigning a name again changes the value it refers to. A result calculated
earlier keeps its existing value until you calculate it again.

In this example, changing `borrowed_books` does not automatically repeat
the earlier calculation of `available_books`:

````python
total_books = 10
borrowed_books = 3
available_books = total_books - borrowed_books

print("Initially available:", available_books)

# Two more books have been borrowed.
borrowed_books = 5

# The earlier result is still stored in available_books.
print("Before recalculating:", available_books)

# Repeat the calculation using the new value.
available_books = total_books - borrowed_books
print("After recalculating:", available_books)
````

Python also allows a name to be reassigned to a value of another type.
For clarity, keep each name's meaning and expected type consistent.

## 5. Convert Between Types

A quoted number is text. Convert it before doing arithmetic.

| Expression | Result | Result type |
| --- | --- | --- |
| `int("12")` | `12` | `int` |
| `float("2.5")` | `2.5` | `float` |
| `str(12)` | `"12"` | `str` |

These conversions require suitable input. For example, `int("hello")`
and `int("2.5")` raise `ValueError` because those strings are not valid
integer input. We will practise handling invalid input in a later chapter.

````python
books_read_text = "12"
study_hours_text = "2.5"

books_read = int(books_read_text)
study_hours = float(study_hours_text)
books_read_label = str(books_read)

print("Books read:", books_read, type(books_read))
print("Study hours:", study_hours, type(study_hours))
print("Books read as text:", books_read_label, type(books_read_label))
````

### Text Addition and Numeric Addition

With strings, `+` joins text. With numbers, `+` performs addition.
Predict the following outputs before running the cell:

````python
first_number_text = "12"
second_number_text = "8"

print("Text result:", first_number_text + second_number_text)
print("Numeric result:", int(first_number_text) + int(second_number_text))
````

### Keep Identifiers as Text

Some values contain digits but act as labels. A student code such as
`"0007"` can contain meaningful leading zeros. Keep it as a string.

Converting it to an integer removes the leading zeros. Converting that
integer back to a string does not recover the original formatting.

````python
student_code = "0007"
numeric_code = int(student_code)

print("Original student code:", student_code)
print("Converted to an integer:", numeric_code)
print("Converted back to text:", str(numeric_code))
````

### A Common Boolean Conversion Mistake

`bool()` checks truthiness. It does not read a text label such as `"False"`
as the Boolean value `False`. A nonempty string is truthy; an empty string
is falsy.

Interpreting text labels requires explicit handling, which we will cover
with conditions in the next chapter.

````python
finished_text = "False"

print("Text type:", type(finished_text))
print('bool("False"):', bool(finished_text))
print('bool(""):', bool(""))
print("Actual Boolean value:", False)
````

## 6. Assignment and Comparison

`=` assigns a value. `==` compares values and returns `True` or `False`.
Here is a small preview of comparisons:

````python
total_lessons = 12
completed_lessons = 3

is_course_complete = completed_lessons == total_lessons

print("Course complete:", is_course_complete)
print("Comparison result type:", type(is_course_complete))
````

## 7. Display Values with f-Strings

Place `f` before a quoted string to create an f-string. Python inserts
the values of names or expressions written inside `{}`.

````python
first_name = "Amina"
age = 24
favourite_language = "Python"

print(f"{first_name} is {age} years old and is learning {favourite_language}.")
````

## 8. Worked Example: A Study Progress Summary

Combine variables, types, arithmetic, and output.
Dividing completed lessons by total lessons gives a ratio. Multiplying
that ratio by 100 gives the percentage.

Here, `total_lessons` is greater than zero, so the division is valid.
A later chapter will show how to handle a total of zero with a condition.

````python
student_name = "Amina"
course_name = "Python Basics"
total_lessons = 12
completed_lessons = 3
target_date = None

remaining_lessons = total_lessons - completed_lessons
progress_ratio = completed_lessons / total_lessons
progress_percent = progress_ratio * 100
is_course_complete = completed_lessons == total_lessons

print(f"Student: {student_name}")
print(f"Course: {course_name}")
print("Total lessons:", total_lessons)
print("Completed lessons:", completed_lessons)
print("Remaining lessons:", remaining_lessons)
print(f"Progress: {progress_percent}%")
print("Course complete:", is_course_complete)
print("Target date:", target_date)
````

## 10. Common Mistakes

| Symptom | Cause | Correction |
| --- | --- | --- |
| `NameError` | A name was used before assignment or spelled differently | Check spelling and run cells in order |
| `TypeError` during subtraction | A value is still text | Convert suitable numeric text first |
| `ValueError` from `int()` | The text is not valid integer input | Inspect the value and choose an appropriate conversion |
| Missing leading zeros | A text identifier was converted to a number | Preserve the original identifier as text |
| A calculated value seems out of date | An input changed after the calculation | Repeat the calculation |
| `bool("False")` returns `True` | A nonempty string is truthy | Handle text labels explicitly |
